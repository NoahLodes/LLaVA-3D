```python
import os
import subprocess
import zipfile  # Import the zipfile module for handling .zip files

# Path to the file containing scan IDs
SCAN_ID_FILE = "/mnt/scratch/LLaVA-3D/data/train_scans.txt"  # Change this if the file is located elsewhere

# Path to the download script
DOWNLOAD_SCRIPT = "/mnt/scratch/LLaVA-3D/data/download3RSCAN.py"  # Ensure this script is in the same directory or provide the full path

# Output directory for downloaded scans
OUTPUT_DIR = "/mnt/scratch/LLaVA-3D/data/3rscan"
```

```python
def extract_zip_files(scan_dir):
    """Extracts all .zip files in the given directory."""
    for root, _, files in os.walk(scan_dir):
        for file in files:
            if file.endswith(".zip"):
                zip_path = os.path.join(root, file)
                extract_path = os.path.join(root, os.path.splitext(file)[0])
                if not os.path.exists(extract_path):
                    print(f"Extracting {zip_path} to {extract_path}...")
                    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                        zip_ref.extractall(extract_path)
                else:
                    print(f"Skipping extraction, already exists: {extract_path}")

def main():
    # Check if the scan ID file exists
    if not os.path.isfile(SCAN_ID_FILE):
        print(f"Error: Scan ID file '{SCAN_ID_FILE}' not found.")
        return

    # Read the scan IDs from the file
    with open(SCAN_ID_FILE, "r") as file:
        scan_ids = [line.strip() for line in file if line.strip()]

    # Download and extract each scan
    for scan_id in scan_ids:
        print(f"Processing scan: {scan_id}")
        scan_dir = os.path.join(OUTPUT_DIR, scan_id)
        try:
            # Run the download script for this scan ID
            subprocess.run(
                ["python3", DOWNLOAD_SCRIPT, "-o", OUTPUT_DIR, "--id", scan_id],
                check=True,
            )
            # Extract any zip files in the scan directory
            extract_zip_files(scan_dir)
        except subprocess.CalledProcessError as e:
            print(f"Failed to process scan {scan_id}: {e}")
```

```python
main()
```
